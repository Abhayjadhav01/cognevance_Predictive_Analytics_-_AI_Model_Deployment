# Retail revenue forecasting exploration

This notebook inspects the reproducible monthly dataset and chronological holdout report. Run `python scripts/acquire_data.py`, `python scripts/prepare_data.py`, and `python scripts/train_model.py` from the project root first. The cleaning policy excludes cancellation invoices and invalid/non-positive quantity or price lines; returns are not netted.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

project_root = Path.cwd()
if not (project_root / 'data').exists():
    project_root = project_root.parent
monthly = pd.read_csv(project_root / 'data/processed/monthly_revenue.csv', parse_dates=['Month'])
metrics = pd.read_csv(project_root / 'reports/model_metrics.csv')
monthly.head(), monthly.describe(), metrics
```

```python
ax = monthly.set_index('Month')['Revenue'].plot(figsize=(12, 4), title='Monthly gross revenue (returns excluded)')
ax.set_ylabel('Revenue')
plt.show()
```

The holdout contains only a few late periods, so report MAE, RMSE, R², and the train/test period counts together. Scores can vary substantially with a different holdout, and negative R² is possible. The recursive forecast is an estimate, not a financial commitment.
